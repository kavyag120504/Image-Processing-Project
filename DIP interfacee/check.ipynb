```python
import torch

# Load your chest model
checkpoint = torch.load("models/best_chest_model.pth", map_location="cpu")

# Try to recover class order
if "class_to_idx" in checkpoint:
    print("Class order found in checkpoint (class_to_idx):")
    class_to_idx = checkpoint["class_to_idx"]
    sorted_by_index = sorted(class_to_idx.items(), key=lambda x: x[1])
    for cls, idx in sorted_by_index:
        print(f"Index {idx} → {cls}")

elif "idx_to_class" in checkpoint:
    print("Class order found in checkpoint (idx_to_class):")
    idx_to_class = checkpoint["idx_to_class"]
    for idx in sorted(idx_to_class.keys()):
        print(f"Index {idx} → {idx_to_class[idx]}")

else:
    print("❌ No class mapping found in model file.")
    print("If you trained using ImageFolder, class order is alphabetical.")

```
